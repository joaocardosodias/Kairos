## 4. Comprehensive Classification Summary Tables

```python
# Create comprehensive classification tables for all datasets
def create_classification_table(classifications, dataset_name):
    table_data = []
    for col_name, details in classifications.items():
        table_data.append({
            'Dataset': dataset_name,
            'Column Name': col_name,
            'Classification': details['type'],
            'Unique Values': details['unique_values'],
            'Explanation': details['explanation'],
            'Preprocessing': details['preprocessing']
        })
    return pd.DataFrame(table_data)

# Create tables for all datasets
table1 = create_classification_table(dataset1_classifications, 'Service Orders')
table2 = create_classification_table(dataset2_classifications, 'Vehicles Base')

# Combine all tables
complete_classification = pd.concat([table1, table2], ignore_index=True)

print("=== COMPLETE CLASSIFICATION TABLE - ALL DATASETS ===")
display(complete_classification)
```

## 5. Classification Summary by Type Across All Datasets

```python
# Analyze classification distribution across all datasets
def categorize_columns(classifications):
    numerical_continuous = []
    numerical_discrete = []
    categorical_nominal = []
    categorical_binary = []
    categorical_ordinal = []
    identifiers = []
    datetime_cols = []

    for col_name, details in classifications.items():
        col_type = details['type']
        if 'Continuous' in col_type:
            numerical_continuous.append(col_name)
        elif 'Discrete' in col_type:
            numerical_discrete.append(col_name)
        elif 'Binary' in col_type:
            categorical_binary.append(col_name)
        elif 'Ordinal' in col_type:
            categorical_ordinal.append(col_name)
        elif 'Identifier' in col_type:
            identifiers.append(col_name)
        elif 'DateTime' in col_type or 'Temporal' in col_type:
            datetime_cols.append(col_name)
        elif 'Categorical' in col_type or 'Nominal' in col_type:
            categorical_nominal.append(col_name)

    return {
        'numerical_continuous': numerical_continuous,
        'numerical_discrete': numerical_discrete,
        'categorical_nominal': categorical_nominal,
        'categorical_binary': categorical_binary,
        'categorical_ordinal': categorical_ordinal,
        'identifiers': identifiers,
        'datetime_cols': datetime_cols
    }

# Categorize columns for each dataset
dataset1_categories = categorize_columns(dataset1_classifications)
dataset2_categories = categorize_columns(dataset2_classifications)

print("=== CLASSIFICATION SUMMARY BY DATASET ===")
datasets = [('Service Orders', dataset1_categories),
           ('Vehicles Base', dataset2_categories)]

for dataset_name, categories in datasets:
    print(f"\n{dataset_name.upper()}:")

    print(f"\n   NUMERICAL:")
    print(f"   Continuous ({len(categories['numerical_continuous'])}): {categories['numerical_continuous']}")
    print(f"   Discrete ({len(categories['numerical_discrete'])}): {categories['numerical_discrete']}")

    print(f"\n   CATEGORICAL:")
    print(f"   Nominal ({len(categories['categorical_nominal'])}): {categories['categorical_nominal']}")
    print(f"   Binary ({len(categories['categorical_binary'])}): {categories['categorical_binary']}")
    print(f"   Ordinal ({len(categories['categorical_ordinal'])}): {categories['categorical_ordinal']}")

    print(f"\n   IDENTIFIERS ({len(categories['identifiers'])}): {categories['identifiers']}")
    print(f"\n   DATETIME ({len(categories['datetime_cols'])}): {categories['datetime_cols']}")

    total_cols = sum(len(cat) for cat in categories.values())
    print(f"\n   TOTAL COLUMNS: {total_cols}")
```

## 7. Data Quality Assessment Across All Datasets

```python
print("=== COMPREHENSIVE DATA QUALITY ASSESSMENT ===")

datasets_info = [
    ('Service Orders', df_service),
    ('Vehicles Base', df_vehicles)
]

quality_summary = []

for dataset_name, df in datasets_info:
    print(f"\n{dataset_name.upper()} QUALITY:")

    for col in df.columns:
        missing_count = df[col].isnull().sum()
        missing_pct = (missing_count / len(df)) * 100

        quality_info = {
            'Dataset': dataset_name,
            'Column': col,
            'Total Records': len(df),
            'Missing Count': missing_count,
            'Missing %': f"{missing_pct:.1f}%",
            'Unique Values': df[col].nunique(),
            'Data Quality': 'High' if missing_pct < 5 else 'Medium' if missing_pct < 20 else 'Low'
        }
        quality_summary.append(quality_info)

        if missing_pct > 10:
            print(f"   {col}: {missing_pct:.1f}% missing")

    completeness = (1 - df.isnull().sum().sum() / (len(df) * len(df.columns))) * 100
    print(f"   Overall Completeness: {completeness:.1f}%")

quality_df = pd.DataFrame(quality_summary)

print("\n=== DETAILED QUALITY SUMMARY ===")
display(quality_df[quality_df['Missing %'] != '0.0%'])  # Show only columns with missing values

print("\nKEY QUALITY FINDINGS:")
high_missing = quality_df[quality_df['Missing %'].str.rstrip('%').astype(float) > 20]
if len(high_missing) > 0:
    print(f" High missing data (>20%): {len(high_missing)} columns")
    for _, row in high_missing.iterrows():
        print(f"     - {row['Dataset']}.{row['Column']}: {row['Missing %']}")
else:
    print(" No columns with high missing data (>20%)")

medium_missing = quality_df[(quality_df['Missing %'].str.rstrip('%').astype(float) > 5) &
                           (quality_df['Missing %'].str.rstrip('%').astype(float) <= 20)]
print(f" Medium missing data (5-20%): {len(medium_missing)} columns")
```
